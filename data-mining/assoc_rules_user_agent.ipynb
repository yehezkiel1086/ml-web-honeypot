```python
import pandas as pd

df = pd.read_json("../tanner_report.json", lines=True)

df.head(5)
```

```python
user_agents = []

for _, row in df.iterrows():
  user_agent = row.get("headers", {}).get("user-agent", "")
  user_agents.append(user_agent)

user_agents
```

```python
user_agents_matrix = []
tmp = []

for i, x in enumerate(user_agents):
  if i % 10 == 0 and i != 0:
    user_agents_matrix.append(tmp)
    tmp = []
  tmp.append(x)

# Add the last group if not empty
if tmp:
  user_agents_matrix.append(tmp)

user_agents_matrix
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpmax, fpgrowth

te = TransactionEncoder()
te_ary = te.fit(user_agents_matrix).transform(user_agents_matrix)
df = pd.DataFrame(te_ary, columns=te.columns_)

df
```

```python
# frequent_itemsets = fpgrowth(df, min_support=0.3, use_colnames=True)
frequent_itemsets = fpgrowth(df, min_support=0.008, use_colnames=True)
### alternatively:
#frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)
#frequent_itemsets = fpmax(df, min_support=0.6, use_colnames=True)

frequent_itemsets
```

```python
import psycopg2

conn = psycopg2.connect(database="web_honeypot_generated", user = "postgres", password = "admin", host = "127.0.0.1", port = "5432")

print("Opened database successfully")
```

```python
# create table
cur = conn.cursor()
cur.execute('''CREATE TABLE assoc_rules_user_agents (
            ID INT PRIMARY KEY     NOT NULL,
            SUPPORT           REAL    NOT NULL,
            USER_AGENT            TEXT     NOT NULL);''')

print("Table created successfully")

conn.commit()
```

```python
# insert data

cur = conn.cursor()

insert_query = """
    INSERT INTO assoc_rules_user_agents (ID, SUPPORT, USER_AGENT)
    VALUES (%s, %s, %s)
"""

for idx, row in frequent_itemsets.iterrows():
    cur.execute(
        insert_query,
        (int(idx), float(row["support"]), str(list(row["itemsets"])))
    )

conn.commit()
print("Records created successfully")
conn.close()
```
