```python
import pandas as pd

df = pd.read_json("../detected.json", lines=True)

df.head(5)
```

```python
attacks = []

for _, row in df.iterrows():
  detection = row.get("response_msg", {}).get("response", {}).get("message", {}).get("detection", {})
  detection_name = detection.get("name", "")
  if detection_name != "index":
    attacks.append(detection_name)

attacks
```

```python
attack_matrix = []
tmp = []

for i, x in enumerate(attacks):
  if i % 10 == 0 and i != 0:
    attack_matrix.append(tmp)
    tmp = []
  tmp.append(x)

# Add the last group if not empty
if tmp:
  attack_matrix.append(tmp)

attack_matrix
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpmax, fpgrowth

te = TransactionEncoder()
te_ary = te.fit(attack_matrix).transform(attack_matrix)
df = pd.DataFrame(te_ary, columns=te.columns_)

df.head(10)
```

```python
# frequent_itemsets = fpgrowth(df, min_support=0.3, use_colnames=True)
frequent_itemsets = fpgrowth(df, min_support=0.001, use_colnames=True)
### alternatively:
#frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)
#frequent_itemsets = fpmax(df, min_support=0.6, use_colnames=True)

frequent_itemsets.head(20)
```

```python
import psycopg2

conn = psycopg2.connect(database="web_honeypot_generated", user = "postgres", password = "admin", host = "127.0.0.1", port = "5432")

print("Opened database successfully")
```

```python
# create table
cur = conn.cursor()
cur.execute('''CREATE TABLE assoc_rules_attack_types (
            ID INT PRIMARY KEY     NOT NULL,
            SUPPORT           REAL    NOT NULL,
            ATTACK_TYPE            VARCHAR(255)     NOT NULL);''')

print("Table created successfully")

conn.commit()
```

```python
# insert data

cur = conn.cursor()

insert_query = """
    INSERT INTO assoc_rules_attack_types (ID, SUPPORT, ATTACK_TYPE)
    VALUES (%s, %s, %s)
"""

for idx, row in frequent_itemsets.iterrows():
    cur.execute(
        insert_query,
        (int(idx), float(row["support"]), str(list(row["itemsets"])))
    )

conn.commit()
print("Records created successfully")
conn.close()
```
