```python
import pandas as pd

df = pd.read_json("../tanner_report.json", lines=True)

df.head(5)
```

```python
paths = df["path"]

# get the paths with params
# filtered_paths = filtered_df["path"]

# get the paths without params
paths = paths.apply(lambda x: x.split('?', 1)[0])

paths_matrix = []
tmp = []

for i, x in enumerate(paths):
  if i % 10 == 0 and i != 0:
    paths_matrix.append(tmp)
    tmp = []
  tmp.append(x)

# Add the last group if not empty
if tmp:
  paths_matrix.append(tmp)

paths_matrix
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(paths_matrix).transform(paths_matrix)
df = pd.DataFrame(te_ary, columns=te.columns_)

df
```

```python
from mlxtend.frequent_patterns import apriori

apriori(df, use_colnames=True, min_support=0.05)
```
